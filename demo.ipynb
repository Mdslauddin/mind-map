```python
import networkx as nx
import matplotlib.pyplot as plt 


# Create a new directed graph 

G = nx.DiGraph()
```

```python
# Add nodes
G.add_node("Central Idea")
G.add_node("Main Topic 1")
G.add_node("Main Topic 2")
G.add_node("Subtopic 1.1")
G.add_node("Subtopic 1.2")
G.add_node("Subtopic 2.1")

# Add edges
G.add_edge("Central Idea", "Main Topic 1")
G.add_edge("Central Idea", "Main Topic 2")
G.add_edge("Main Topic 1", "Subtopic 1.1")
G.add_edge("Main Topic 1", "Subtopic 1.2")
G.add_edge("Main Topic 2", "Subtopic 2.1")

# Draw the graph
pos = nx.spring_layout(G)
nx.draw_networkx(G, pos, with_labels=True, node_color='lightblue')
plt.show()
```
